## 3. การตรวจสอบ Missing value

```python
df[["Player","Rk","Pos"]].isna().sum()
```

```python
df["PTS"] = pd.to_numeric(df["PTS"],errors = "coerce")
df["FT%"] = pd.to_numeric(df["FT%"], errors = "coerce")
df["3PA"] = pd.to_numeric(df["3PA"], errors = "coerce")
df["3P%"] = pd.to_numeric(df["3P%"], errors ="coerce")
```

```python
df[df["3P%"].isna()]
#แสดงผู้เล่นที่มีค่างว่างใน column 3P%
#การคำนวน 3 เเต้ม 3P / 3PA
```

```python
df[df["3P%"].isna()][["Player","3P","3PA","3P%"]]
# missing value ของคอลัมน์ 3P% ไม่ได้เกิดจากข้อมูลสูญหาย เเต่เกิดจากไม่สามารถคำนวนค่าได้ (3P/3PA = 3P%)
```

```python
df[df["FG%"].isna()]
# FG/FGA = FG%
df[df["FG%"].isna()][["Player","FG","FGA","FG%"]]
# ผู้เล่นไม่ได้ทำเเต้มทำให้ไม่มีค่า FG%(จำนวนลูกที่ยิงลง)
```

```python
df[df["2P%"].isna()][["Player","2P","2PA","2P%"]]
# missing value เกิดจากไม่สามารถคำนวนค่าได้
```

```python
df[df["eFG%"].isna()][["Player","FG","FGA","3P","eFG%"]]
# eFG = (FG+0.5(3P))/FGA missing value เกิดจากไม่สามารถคำนวนค่าได้
```

```python
df[df["FT%"].isna()][[ "Player","FT","FTA","FT%"]]
# FT% = FT/FTA missing value เกิดจากการไม่สามารถคำนวนค่าได้
```

## 5. Insights ที่ได้จากการวิเคราะห์


ตำแหน่งที่มี PTS เฉลี่ยสูงสุด
จากข้อมูลพบว่าตำเเหน่งที่มีค่าเฉลี่ย PTS สูงสุดคือ PG ซึ่งมีค่าเฉลี่ยอยู่ที่ 8.64 คะเเนนต่อเกม ขณะที่ SG มีค่า PTS เฉลี่ยประมาณ 8.12 คะเเนนต่อเกม ซึ่งเเสดงให้เห็นว่า ในข้อมูลชุดนี้มีเเนวโน้มว่าตำเเหน่ง PG จะมีค่า PTS เฉลี่ยสูงกว่าผู้เล่นตำเเหน่ง SG

ผู้เล่นกลุ่มอายุไหนที่มีคะเเนนเฉลี่ยสูงกว่า
จากข้อมูลพบว่าผู้เล่นที่มีอายุ 25 ปีขึ้นไป มีค่า PTS เฉลี่ยต่อเกมอยู่ที่ประมาณ 8.64 คะเเนนต่อเกม ซึ่งมากกว่าคะเเนน PTS เฉลี่ยของผู้เล่นที่มีอายุน้อยกว่า 25 ปีอยู่ที่ประมาณ 7.19 คะแนนต่อเกม

พบความสัมพันธ์ระหว่างอะไรบ้าง
จากข้อมูล ค่าสหสัมพันธ์ระหว่าง Assist(AST) เเละ Turnover(TOV) อยู่ที่ 0.8313 ซึ่งเเสดงให้เห็นว่าตัวเเปรทั้งสองมีความสัมพันธ์เชิงบวกค่อนข้างสูง โดยในชุดข้อมูลนี้ ผู้เล่นที่มีค่า Assist สูงมีเเนวโน้มที่จะมีค่า Turnover สูงตามไปด้วย
